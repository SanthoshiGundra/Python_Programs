```python
# write a python program that prints each line in the file in reverse order
with open("Python.txt","r") as f:
  lines = f.readlines()
  for line in lines:
    print(line[::-1])
```

```python
# write a python program that prints each line in the file in reverse order
with open("Python.txt","r") as f:
  lines= f.readlines()
  for line in reversed(lines):
    print(line.rstrip())
```

```python
# write a program to sort words in a file and put them in another file. The output file should be only lower case words so any upper case words are removed
with open("Python.txt","r") as f:
  data = f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
  print(sort_data)
  with open("SORT.txt","w") as f2:
    for i in sort_data:
      f2.write(i+"\n")
  print("Done")
```

```python
# python program to compute the number of characters, words and lines in a file
with open("Python.txt","r") as f:
  lines=f.readlines()
  print(len(lines))
  words=0
  for line in lines:
    words+=len(line.split())
  print(words)
  chars=0
  for line in lines:
    chars+=len(line)
  print(chars)
```

```python
import pickle as p
with open("s1.bin","wb") as f: #.bin or .dat
  L=[1,2,3,"Hello"]
  p.dump(L,f)   #writing: dump
  print("Done") #reading: load
  p.dump("World",f)
  print("Done")
with open("s1.bin","rb") as f1:
  print(p.load(f1))
  print(p.load(f1))
  print("Done")
```
