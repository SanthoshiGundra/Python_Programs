# **SETS IN PYTHON**

```python
# INITIALISATION
s={}
a=set()
b={1,2,3,"new"}
print(type(s),type(a),type(b))
print(s)
print(a)
print(b)
```

```python
# SET OPERATIONS
a={1,2,3,4,5}
b={3,4,6,"hi"}
print(a|b) #union
print(a&b) #intersection
print(a-b) #difference
print(a^b) #symmetric difference
```

```python
# SET METHODS
a={1,2,3,4,5}
b={3,4,6,"hi"}
print(a.union(b))
print(a)
print(a.intersection(b))
print(a)
print(a.difference(b))
print(a)
print(a.symmetric_difference(b))
print(a)
# Fix 1: 'set' object has no attribute 'union_update'. Use 'update()' instead.
# 'update()' modifies the set in-place and returns None.
a.update(b)
print(a)
# These 'print(a._update(b))' lines will print None, then the next print(a) will show the modified set.
# It's usually better to call the method and then print the set separately.
a.intersection_update(b)
print(a)
a.difference_update(b)
print(a)
a.symmetric_difference_update(b)
print(a)
b.add("python")
print(b)
# Fix 2: 'update()' expects a single iterable argument.
a.update({6,7})
print(a)
a.discard(6)
print(a)
```
