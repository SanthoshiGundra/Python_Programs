```python
# TUPLE OPERATION
t=(1,2,3,4,15,8,68,38)
t1=("hello","hi","python","32",45,67)
print(t+t1) #concatenation
print(t*2) #repetition
print(t[4]) #indexing
print(t1[-4]) #reverse indexing
print(t1[:5]) #slicing
print(t1[3:])
print(t1[2:5])
#membership operator
if "python" in t1:
    print("yes")
else:
    print("no")
#iteration
for i in t1:
    print(i,end=" ")
print()
for i in range(len(t)):
    print(t[i],end=" ")
```

```python
# ZIP FUNCTION
t=("n1","n2","n3")
t1=(1,2,3)
t2=("hello",23,88,"python")
print(zip(t,t1))
print(tuple(zip(t,t1)))
print(tuple(zip(t,t2)))
```

```python
# BUILT-IN FUNCTIONS
t=(1,2,3,4,15,8,68,38)
t1=("hello","hi","python","32",45,67)
print(len(t))
print(len(t1))
print(max(t))
print(min(t))
print(sorted(t))
print(sum(t))
print(sum(t)/len(t))
print(reversed(t))
print(reversed(t1))
print(tuple(reversed(t)))
print(tuple(reversed(t1)))
t3=('a','b','c','e','d')
print(sorted(t3))
print(sorted(t3,reverse=True))
print(sorted(t3,key=str.lower))
print(sorted(t3,key=str.upper))
```
