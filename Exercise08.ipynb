```python
# write  a python program to count no.of duplicates in given data
L=[int(i) for i in input().split()]
s=set(L)
print(s)
for num in s:
  count=L.count(num)
  if(count>1):
    print(num ,end=" ")
```

```python
#write a python program to print non repeated elements
L=[int(i) for i in input().split()]
s=set(L)
print(s)
for num in s:
  count=L.count(num)
  if(count==1):
    print(num ,end=" ")
```

```python
# write a python program that prints missing numbers between 1 to n
n=int(input())
s={int(i) for i in input().split()}
print(s)
S={int(j) for j in range(1,n+1)}
print(S)
print(S-s)
```
