```python
#progarm that prints reverse a number using a function
def reverse(num):
  rev=0
  while(num>0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  return rev
print(reverse(12345))
print(reverse(987654321))
print(reverse(0))
```

```python
#program that reverse a string using user defined functions
a=input()
def reverse(str):
  rev=""
  for i in str:
    rev=i+rev
  return rev
print(reverse(a))
```

```python
#program that checks whether string is palindrom or not
def palindrome(s):
  a=s[::-1]
  if(s==a):
    print("String is Palindrome")
  else:
    print("String is not a Palindrome")
t=input()
palindrome(t)
```

```python
# A function that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg = multiple_function(1,2,3,4,5,6,7,8,9)
print("The sum is",res_sum,"\nThe count is",res_count,"\nThe average is",res_avg)
```
