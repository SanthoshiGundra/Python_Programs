```python
#program on Arithematic operators
x=int(input("enter a value: "))
y=int(input("enter another value:"))
print("the sum of the inputs is",x+y)
print("subtracting x from y gives",x-y)
print("the product of the inputs is",x*y)
print("the quotient when x is divide by y is",x/y)
print("the remainder when x is divided by y is",x%y)
print("floor division outputs",x//y)
print("x to the power of y is",x**y)
```

```python
#Assignment operators
y=5
x=2
print("the value of x is",x,"and the value of y is",y)
```

```python
#Comparision operator
print(x==y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
print(x!=y)
```

```python
#logical operator
print(x>y and x!=y)
print(x<y or x==y)
print(not(x==5) )
```

```python
#identity operator
print(x is not y)
print(x is y)
```

```python
#Bitwise operators
print(x & y)
print(x|y)
print(x^y)
print(x<<2)
```
