```python
a="hello world"
print("a is ",a)
b=a.split() #used to remove the spaces or selected characters
print("b is ",b)
c="@".join(b) #join a character to the string whenever a space occurs
print("c is ",c)
print(c.count('l')) #count the repetition of letter in the string
```

```python
print("hi hello namaste".split())
dept='   Ai&Ds   '
print(dept)
print(dept.strip()) #removing head and tail spaces
print(dept.replace("Ds","Ml")) #used to replace words/letters in strings
print(dept.find("Ds")) #can find the index of starting letter
print(dept.lower()) #all letters in the string into lower case
print(dept.upper()) #changing case to upper
```

```python
s="Hello World"
t=" welcome to python classes"
print(s+t) #concatenation
print(s*2) #repetition
print(s[6]) #forward indexing
print(t[-6]) #backward indexing
print(s[1:11]) #forward slicing
print(t[-6:-1]) #backward slicing
print(s[::-1]) #reversing
if 'a' in t:
  print("yes") #membership
for i in s:
  print(i) #iteration
```
