# **DICTONARIES IN PYTHON**

```python
#dictionary comprehension
k={i: i*i for i in range(1,6)}
print(k)
```

```python
#user defined dictonary
k=[i for i in input().split()]
v=[int(i) for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)
```

```python
#write a program to check if a given key exists  in a dictionary or not
d={"sno":110,"name":"Ram","grade":10.00,"rank":1};
print(d);
key=input();
if key in d:
    print("key exists");
else:
    print("key does not exist");
```

```python
#write a program to add a new key value pair to an existing dictionary
d={"sno":110,"name":"Ram","grade":10.00,"rank":1}
print(d)
d.update({"class":10})
print(d)
```

```python
#write a program that prints frequency or occurence of each element in a list in the following format
#format: input=[1,2,3,4,1,5,6,1,6,7]  output={1:3,2:1,3:1,4:1,5:1,6:2,7:1}
n=list(map(int,input().split()))
d={}
for i in n:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
#write a program that prints frequency or occurence of each element in a string in the following format
s=input().split()
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```
